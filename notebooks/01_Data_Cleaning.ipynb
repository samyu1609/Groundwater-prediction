```python
print("Groundwater")
print(groundwater.columns)

print()

print("Weather")
print(weather.columns)

print()

print("Rainfall")
print(rainfall.columns)
```

```python
weather_data = pd.merge(
    weather,
    rainfall,
    on="DATE",
    how="outer",
    suffixes=("_NASA", "_OpenMeteo")
)

weather_data.head()
```

```python
final_data = pd.merge(
    groundwater,
    weather_data,
    on="DATE",
    how="left"
)

final_data.head()
```

```python
final_data.isnull().sum()
```

```python
final_data["YEAR"] = final_data["DATE"].dt.year
final_data["MONTH"] = final_data["DATE"].dt.month
final_data["DAY"] = final_data["DATE"].dt.day
```

```python
def season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Post Monsoon"

final_data["SEASON"] = final_data["MONTH"].apply(season)
```

```python
print(final_data.shape)

final_data.head()
```

```python
final_data.to_csv(
    "../data/processed/final_dataset.csv",
    index=False
)

print("Final dataset saved successfully!")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
```

```python
# Linear Regression
lr = LinearRegression()
lr.fit(X_train, y_train)

# Random Forest
rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
rf.fit(X_train, y_train)

# XGBoost
xgb = XGBRegressor(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    random_state=42
)
xgb.fit(X_train, y_train)
```

```python
models = {
    "Linear Regression": lr,
    "Random Forest": rf,
    "XGBoost": xgb
}

results = []

for name, model in models.items():

    pred = model.predict(X_test)

    mae = mean_absolute_error(y_test, pred)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    r2 = r2_score(y_test, pred)

    results.append([name, mae, rmse, r2])

import pandas as pd

results = pd.DataFrame(
    results,
    columns=[
        "Model",
        "MAE",
        "RMSE",
        "R2 Score"
    ]
)

results
```
